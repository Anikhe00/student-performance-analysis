### Data Loading and Initial Inspection

- Import the necessary libraries
- Load the dataset
- Inspect the first few rows

```python
#Import all necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# A dictionary of the categorical columns and their types
cat_dtypes = {"Gender": "category", "ParentalSupport": "category", "Online Classes Taken": "category"}

# Import the data and set "dtype" parameter to "cat_dtypes"
students_df = pd.read_csv("../data/processed/student_performance_cleaned.csv", dtype=cat_dtypes)

# Inspect dataset
print(f"This dataset has {students_df.shape[0]} rows and {students_df.shape[1]} columns.")
students_df.head()
```

```python
# Confirm data types and check for any missing values
print(students_df.dtypes)
print(students_df.isnull().sum())
```
